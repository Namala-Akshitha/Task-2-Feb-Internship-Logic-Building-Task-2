```python
#Problem 1:Unique Words in a Sentence
sentence=input('enter any sentence:')
#splits the sentence into words
words=sentence.split()
#converts list into set(removes duplicates)
unique_words=set(words)
print('Unique words count:',len(unique_words))
print('Unique words:',unique_words)
```

```python
#Problem 2:Highest Salary from Employee Data
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}
#Values represent employee salaries
highest_salary = max(employees.values())
#Keys represent employee names
emp_name=max(employees,key=employees.get)
print('Highest Salary:',emp_name,'-',highest_salary)
```

```python
#Problem 3: Find Maximum and Minimum Values
li =  [45, 22, 89, 10, 66]
# Initialize max and min with the first element
maximum = li[0]
minimum = li[0]
# Loop through the list
for num in li:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num
# Display results
print("List:", li)
print("Max:", maximum)
print("Min:", minimum)
```

```python
#Problem 4: Count Products Above a Price Threshold
prices=[450, 1200, 899, 1500, 300]
count=0  #Start from 0
for p in prices:
    if p>=1000:
        count=count+1 #it increases whenever the condition becomes true.
print("Price above 1000:",count)  #finally displays count
```

```python
#Problem 5: Calculate Attendance Percentage
attendance=["P", "P", "A", "P", "P"]
#Count number of Presents
present_count = attendance.count("P")
#Total days
total_days = len(attendance)
#Calculate percentage
percentage = (present_count/total_days)*100
print("Attendance Percentage:", percentage)
```

```python
#Problem 6: Remove Duplicate Phone Numbers
phone_numbers = [9876543210, 9123456789, 9876543210]
unique_phone=set(phone_numbers)
print('Unique phone numbers:',unique_phone)
```

```python
#Problem 7: Count Character Frequency
string="ppython"
# Create empty dictionary
char_count={}
#Count each character
for ch in string:
    if ch in char_count:
        char_count[ch]+=1
    else:
        char_count[ch]=1
print(char_count)
```
